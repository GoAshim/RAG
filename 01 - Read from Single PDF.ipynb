```python
# Import necessary libraries
import os
from dotenv import load_dotenv

from llama_index.llms.openai import OpenAI
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader
```

```python
# Get API key from environment variable
load_dotenv()
OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

# Validate that the API key is set
if not OPENAI_API_KEY:
    raise ValueError("OPENAI_API_KEY is not set in the environment variables.")
```

```python
# Initialize the OpenAI LLM with the API key and model
OpenAI(api_key=OPENAI_API_KEY, model="gpt-4.0-mini")

# Load documents from the specified directory
documents = SimpleDirectoryReader("pdf").load_data()

# Create a in memory vector store index from the loaded documents
index = VectorStoreIndex.from_documents(documents)

# Query the index with a question
query_engine = index.as_query_engine()

# Example query
question = "How many questions are there in the AWS Certified AI Practitioner exam?"

# Get the response from the query engine
response = query_engine.query(question)

# Print the response
print(response)
```
